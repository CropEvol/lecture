# 遺伝的変異について

遺伝的変異とは、生物集団内や個体間で見られるDNA配列の差異のことを指します。

この差異によって、生き物の持つ表現型に違いが生み出されています。

例えばイネの"こしひかり"と"あきたこまち"は、どちらもイネなのでDNA配列は殆ど同じですが、

一部のDNA配列が異なっていることで味や見た目が異なっています。

つまり、"こしひかり"と"あきたこまち"のDNA配列を比較し、違いを見つける事が出来れば味や見た目の違いを生み出す原因遺伝子等を見つける事が出来る、ということになります。

<img src="https://github.com/CropEvol/lecture/blob/master/images/exp_2025_2/mutation2.png?raw=true" alt="title" height="200px">

この様な形質の違いをスタートに原因となる遺伝子を探索するアプローチを**順遺伝学**と呼びます。

逆にゲノム編集等で遺伝子を壊して形質がどう変化するか見るアプローチを**逆遺伝学**と呼びます。

<img src="https://github.com/CropEvol/lecture/blob/master/images/exp_2025_2/reverse_genetics.png?raw=true" alt="title" height="400px">

今回はこの順遺伝学を進めるために必須となる遺伝的変異の検出方法の実習をまず行います。


# NGS(次世代シーケンサー)による変異検出

遺伝的変異を検出するためには個体ごとにDNA配列を読み取って比較する必要があります。

この時、次世代シーケンサーと呼ばれる技術を使用した**ショートリード**というものを利用することが多いです。

ショートリードは現在最もよく使用されるシーケンスになり、100~300塩基ほどの長さのリードを安価で大量に読む事が出来ます。

前期の実習では、ある特定の遺伝子の配列のみを読み取りましたが、今回はゲノム全体の配列を読み取ります。

<img src="https://github.com/CropEvol/lecture/blob/master/images/exp_2025_2/short_read.png?raw=true" alt="title" height="300px">

次世代シーケンサーでは大量の配列データが出力されるのですが、いずれもDNA配列の断片の情報しか持っていないので、このままだと使い道がありません。

そこで、既にゲノムが完全に解読された配列と比較をします。

モデル生物や主要な作物等は全ゲノムが解読されており、リファレンスゲノムとして[NCBIのGenomes](https://www.ncbi.nlm.nih.gov/home/genomes/)や[Ensemble](https://www.ensembl.org/)などのページで公開されています。

今回扱うイネの場合"**日本晴れ**"等の系統の全ゲノム配列が公開されています。

参考: https://www.ncbi.nlm.nih.gov/datasets/genome/GCF_034140825.1/

この大量に読んだショートリードをリファレンスゲノムに対し貼り付けていき(**マッピング**)、配列の異なる場所を探すことで遺伝的変異の情報を取得します。

<img src="https://github.com/CropEvol/lecture/blob/master/images/exp_2025_2/short_read_mapping.png?raw=true" alt="title" height="500px">

例えば"こしひかり"のショートリードを手に入れた場合は、"日本晴れ"のリファレンスゲノムにマッピングすることで、

* 6番染色体の230,492 bp付近がTからGに変異している.

といった情報が手に入ります。

これが遺伝的変異を検出する一般的な手法になります。

#### 配列、リードのデータファイル形式（FASTA, FASTQ）

##### FASTA

　まず、配列情報は**FASTA**と呼ばれる形式でファイルに保存されています。FASTA形式は主に2つの要素で構成されています。

```
>chromosome1
ATGCACTGACATGCATGACATAGCTAGCATAGGGGCAGT
GTCATACGATCGTACGACTCGTAGACTACGCATCGATCA
ATCAGTACGACGACGATCGATCGACTCGATCAGCTACGA
>chrmosome2
ATGCTAGGCTAGATCGATACTCCCATCATATACATTACA
...
```
- ヘッダー行（>からはじまる行）: 配列のIDや生物種、解説など
- 2行目以降に、塩基（A, T, G, C, Uなど）やアミノ酸を並べる。途中の改行は自由。

##### FASTQ

　シーケンスしたリードのデータは**FASTQ**と呼ばれる形式でファイルに保存されています。FASTQ形式では、4行で1つのリードデータの情報を表しています。

```
@リードの名前
TGAAGCAGT..............(リードの塩基配列)
+
F1GCFCGGE..............(各塩基のクオリティ)
```

- 第1行（@からはじまる行）: リードの名前
- 第2行: リードの塩基配列
- 第3行（+からはじまる行）: 第2行と第4行の区切りのための行
- 第4行: 各塩基配列のクオリティ（ASCIIコードで書かれている）。1文字が何らかのクオリティスコア（数値）を示している。
 - 参考: https://bi.biopapyrus.jp/rnaseq/qc/fastq-quality-score.html

### 検出した変異のフィルタリング

こうして見ると、片方のサンプルの遺伝子型情報が無かったり、あったとしても張り付いたリード数が少なく、信頼性が低い場所などがあることがわかります。

<img src="https://github.com/CropEvol/lecture/blob/master/images/exp_2025_2/vcf_qual.png?raw=true" alt="title" height="150px">

そこで、通常は作成したVCFファイルをそのまま使用するのではなく、**VCFのフィルタリング**という作業を行います。

得られた変異のデータの中で、信頼性の高い場所の情報だけ取り出す作業です。

このVCFのフィルタリングには、`vcftools`というプログラムを使います。
(`bcftools`ではなく、`vcftools`です。)

下記コマンドで`vcftools`をインストールします。

```
apt install vcftools
```



#### **リードのデプスが基準を下回る変異を削除**

何本のリードによって遺伝子型が検出されたのか、この張り付いたリードの数をデプスと呼びます。

デプスが大きければ信頼性が高く、小さければシーケンスエラーの可能性が高くなります。

そこで、このデプスが5以上の部分だけ取り出す、みたいな事が出来ます。

```
vcftools --vcf 入力VCFファイル名 --minDP 5 --recode --out minDP5
```

実行すると`minDP5.recode.vcf`というファイルが出力されます


#### **Genotyping率が基準を下回る変異を削除**

今回は2サンプル(こしひかり&あきたこまち)なので、両サンプル(100%)遺伝子型が決まっている場所か、1サンプルだけ(50%)遺伝子型が決まっている場所か、という風に場合分けする事が出来ます。

集団の解析をしていると、集団の30%の個体しか遺伝子型が決まっていない場所や集団の90%の個体の遺伝子型が決まっている場所など、

各変異ごとに、どれだけの個体について遺伝子型が決まっているか異なります。

<img src="https://github.com/CropEvol/lecture/blob/master/images/exp_2025_2/missing_SNP.png?raw=true" alt="title" height="100px">

そこで、集団中のある程度の個体の遺伝子型が決まっている場所だけ抽出する、という事が出来ます。

集団の80%の個体がちゃんと遺伝子型が決まっている場所を取り出す場合は、

```
vcftools --vcf 入力VCFファイル名 --max-missing 0.8 --recode --out minDP5_mm0.8
```

実行すると`minDP5_mm0.8.recode.vcf`というファイルが出力されます。

この様な形で得られた遺伝的変異の中でもより信頼性の高い情報のみを取り出していきます。

他にも、VCFファイルのクオリティスコアを基準に選抜したりします。

### QTL解析のプログラム

使用する統計モデルや考慮する遺伝的効果の種類などに応じてプログラムを選択する必要があります。

````python
# hailを動かすPythonプログラム
import hail as hl
from hail.plot import show

# 初期設定
hl.plot.output_notebook()

# # リファレンスゲノムの設定
# hl.genetics.ReferenceGenome.from_fasta_file(name="reference", fasta_file="assembly.fasta", index_file="assembly.fasta.fai")

# リファレンスゲノムの設定
contigs = {"chr01": 43270923, "chr02": 35937250, "chr03": 36413819, "chr04": 35502694, "chr05": 29958434, "chr06": 31248787, "chr07": 29697621, "chr08": 28443022, "chr09": 23012720, "chr10": 23207287, "chr11": 29021106, "chr12": 27531856}
lengths = {"chr01": 43270923, "chr02": 35937250, "chr03": 36413819, "chr04": 35502694, "chr05": 29958434, "chr06": 31248787, "chr07": 29697621, "chr08": 28443022, "chr09": 23012720, "chr10": 23207287, "chr11": 29021106, "chr12": 27531856}

# Create the custom reference genome
my_rg = hl.ReferenceGenome(
    name="reference",
    contigs=list(contigs.keys()),
    lengths=lengths,
)

# VCFファイルの読み込み
# hl.import_vcf('minDP5_mm0.8.recode.vcf', reference_genome="reference").write('minDP5_mm0.8.recode.mt', overwrite=True)
hl.import_vcf('SHONI.vcf', reference_genome="reference").write('SHONI.mt', overwrite=True)
mt = hl.read_matrix_table('SHONI.mt')

# 形質データの読み込み
table = hl.import_table('Phenotype.txt', impute=True).key_by('Line')
mt = mt.annotate_cols(pheno=table[mt.s])

# QTL解析の実行
qtl_result = hl.linear_regression_rows(y=mt.pheno.Phenotype, x=mt.GT.n_alt_alleles(), covariates=[1.0])

# 結果をtsvファイルとして出力
qtl_result.export("qtl_result.tsv")

# Manhattan plotの描写
p = hl.plot.manhattan(qtl_result.p_value)
show(p)
````

上手く結果が出ていれば、11番染色体の末端付近にピークが出ているかと思います。

具体的な数値データは`qtl_result.tsv`として出力されています。

最もp値の低かった変異の遺伝子型と形質の関係性をBoxplotで表示してみると、関係性が良く分かります。

````python
import hail as hl
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

target_locus = "chr02:25651669" # ここをピークの出た位置に変える

target_locus = hl.parse_locus(target_locus, reference_genome='reference')
target_mt = mt.filter_rows(mt.locus == target_locus)

entries_table = target_mt.entries()
plot_df = entries_table.select(
    genotype=entries_table.GT.n_alt_alleles(),
    phenotype=entries_table.pheno
).to_pandas()

plot_df = plot_df.dropna(subset=['genotype', 'phenotype.Phenotype'])

plt.figure(figsize=(6, 5))
sns.boxplot(x='genotype', y='phenotype.Phenotype', data=plot_df)
sns.stripplot(x='genotype', y='phenotype.Phenotype', data=plot_df, color='black', alpha=0.3, jitter=0.2)

plt.xlabel('Number of Alternative Alleles (Genotype)')
plt.ylabel('Phenotype Value')
plt.xticks([0, 1, 2], ['Hitomebore', 'Hetero', 'SHONI'])
plt.show()
````

ちなみに11番染色体の長さは大体29M(29000000)bpです

````python
# startからendまでの間にいる遺伝子をリストアップ
import pandas as pd

start = 0
end = 100000

annotation = pd.read_csv("SHONI_chr11_annotation.gff", sep="\t", comment="#", header=None)
annotation.columns = ["seqid", "source", "type", "start", "end", "score", "strand", "phase", "attributes"]
annotation[(annotation.type == "gene") & (annotation.start >= start) & (annotation.end <= end)]
````
