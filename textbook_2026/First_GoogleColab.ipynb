# Google Colaboratoryの使い方


　「ゲノム情報解析入門」では、**Google Colaboratory**（以下、**Google Colab**）と呼ばれるオンラインサービスを使って、ゲノム解析の体験をしていきます。今回の実習では、Google Colabの使い方を勉強していきます。


## Google Colaboratory
　Google ColabはGoogle社が提供しているサービスです。おもにPythonプログラミングをおこなう場として提供されています。ゲノム解析用のソフトウェアをインストールすれば小規模なゲノム解析もおこなえるため、この講義ではGoogle Colabを利用して実習をおこないます。


## ゲノム解析で使う技術

　ゲノム解析のためのソフトウェアは、日々開発されており、今では非常に充実しています。たいていの場合、自身がおこないたいゲノム解析に対応するソフトウェアがすでに存在しているため、そのソフトウェアを実行できればゲノム解析をおこなえる時代です。そのため、ゲノム解析では、ゲノム解析用のソフトウェアを実行するための技術を最もよく使用します。

　ゲノム解析用のソフトウェアの多くは、マウスによるクリック等の操作ではなく、コマンドと呼ばれる文字によるパソコンへの命令文で実行できるように作られています。この講義の実習でも、コマンドを使ったソフトウェアの実行などをおこないます。

### 5-2. 塩基配列をアミノ酸配列に翻訳する

　遺伝子の塩基配列は、3塩基（**コドン**）ごとに1つのアミノ酸に対応しています。どのコドンがどのアミノ酸に対応するかを定めたものを**遺伝暗号**（**コドン表**）と呼びます。

　ここでは、先ほどダウンロードしたヒトのミトコンドリアゲノムから、**COX1遺伝子**（シトクロムc酸化酵素サブユニット1、呼吸に関わる酵素の遺伝子）の領域を切り出し、アミノ酸配列に翻訳してみましょう。COX1遺伝子は、ミトコンドリアゲノムの5904〜7445番目の塩基に位置しています。

````
from Bio import SeqIO
seq = next(SeqIO.parse("Human_mt.fasta", "fasta")).seq

## COX1遺伝子の領域（5904〜7445番目の塩基）を切り出す
## ※ Pythonでは位置を0から数えるため、[5903:7445] と書く
cox1 = seq[5903:7445]
print("長さ:", len(cox1), "bp")
print("コドン数:", len(cox1) // 3)

## 一般的な遺伝暗号表（標準遺伝暗号, table=1）で翻訳する
aa_std = cox1.translate(table=1)
print(aa_std)
print("終止コドン（*）の数:", aa_std.count("*"))
````

　アミノ酸配列の表では、`*` は「翻訳終止」を表していました。本来、1つのタンパク質の配列では、終止コドン `*` は**最後に1つだけ**現れるはずです。ところが、上の結果では配列の途中に `*` が何個も現れています。これはなぜでしょうか？

　実は、ミトコンドリアは、核の遺伝子とは**少し異なる遺伝暗号**を使っています。Biopythonでは `table=2` を指定すると、脊椎動物のミトコンドリアの遺伝暗号で翻訳できます。

### 練習
　次のコードセルの `table=` の後に `2` を書き入れて実行し、終止コドン `*` の数がどう変わるか確認してください。

````
## 脊椎動物ミトコンドリアの遺伝暗号表（table=2）で翻訳する
aa_mt = cox1.translate(table=)
print(aa_mt)
print("終止コドン（*）の数:", aa_mt.count("*"))
````

　どのコドンの読み方が違ったのかを調べてみましょう。下のコードセルは、2つの遺伝暗号表で翻訳結果が異なるコドンを数えるプログラムです。

````
from collections import Counter
diff = Counter()
for i in range(0, len(cox1), 3):
    codon = cox1[i:i+3]
    a1 = str(codon.translate(table=1))  # 標準遺伝暗号
    a2 = str(codon.translate(table=2))  # ミトコンドリアの遺伝暗号
    if a1 != a2:
        diff[(str(codon), a1, a2)] += 1

print("コドン  標準  ミトコンドリア  出現回数")
for (codon, a1, a2), n in diff.items():
    print(f"{codon}     {a1}     {a2}             {n}")
````

　結果から、ヒトのミトコンドリアでは次のような違いがあることがわかります。

|コドン|標準遺伝暗号|ミトコンドリアの遺伝暗号|
|:--|:--|:--|
|TGA|終止（*）|トリプトファン（W）|
|ATA|イソロイシン（I）|メチオニン（M）|
|AGA|アルギニン（R）|終止（*）|

　遺伝暗号は「すべての生物で共通」と習うことが多いですが、実際にはミトコンドリアや一部の生物で例外があります。NCBIなどのデータベースでも、遺伝子ごとに使う遺伝暗号表の番号（`transl_table=2` など）が記録されています。

<small>※ プログラム上は `table=1` と `table=2` の違いだけですが、この違いを知らずに解析すると、まったくおかしな結果になってしまいます。後の「7. 生成AIの利用について」で扱う「背景知識の重要性」の一例です。</small>

### 5-3. FASTQ形式
　もうひとつは、**FASTQ**形式とよばれる記述様式で、先ほど勉強したFASTA形式とは異なり、塩基配列の名前と塩基配列本体だけでなく、各塩基のクオリティ（信頼度）も記述できる形式です。

```
@read1 ←塩基配列の名前
TGCTGCGGAGCCCACGCCACGATCTTCCCGAACCCCTTCG ←塩基配列の本体
+ ←塩基配列本体とクオリティの区切り
JJJJJJJJJJJ<JJJJJJJJJJJJJJJJJJJJJJJJJJJA ←各塩基のクオリティ
@read2
TTTCTCCCCATCAATCGACAATATTTCTTGCTTTCTCGCC
+
7JJJFFJJFJJJJJFJFJJFJ-JJJJFJJFFFJJJJJJJJ
@read3
CAGCGCCTTCTCTTCTATAAGCACCCCTGCACCCCTACTC
+
<FJJJ<JJ<JJ<JJJJJJJ<F-FJ-7FJ7-FJJJJFAFJF
```

　FASTQ形式は、次世代シーケンサーや第3世代シーケンサーから得られる塩基配列データでよく使われています。

### 練習: クオリティの文字を数値に変換する

　FASTQ形式のクオリティは、1文字で1塩基分のスコアを表しています。コンピュータの中では、文字は番号（ASCIIコード）で管理されていて、`!` が33番、`"` が34番、…と続きます。そのため、**文字の番号から33を引く**とPhredクオリティスコアになります。

　次のコードセルを実行して、上の例の `read1` のクオリティを数値に変換してみましょう。

````
qual = "JJJJJJJJJJJ<JJJJJJJJJJJJJJJJJJJJJJJJJJJA"

## ord() は文字の番号（ASCIIコード）を返す
scores = [ord(c) - 33 for c in qual]
print(scores)

## Phredクオリティスコア Q から、塩基が不正確である確率を計算する: 10^(-Q/10)
for q in sorted(set(scores)):
    print(f"Q={q}: 不正確な確率 {10**(-q/10):.5f}")
````

<small>※ 上の表は I（Q=40）までですが、最近のIlluminaのシーケンサーでは J（Q=41）まで使われることがあります。</small>

　次のコードセルの `qual` を、上の例の `read2` のクオリティ（`7JJJFFJJFJJJJJFJFJJFJ-JJJJFJJFFFJJJJJJJJ`）に書き換えて実行し、**最もスコアの低い塩基**と、その塩基が不正確である確率を調べてください。

````
qual = "JJJJJJJJJJJ<JJJJJJJJJJJJJJJJJJJJJJJJJJJA"  # ← read2のクオリティに書き換える

scores = [ord(c) - 33 for c in qual]
q_min = min(scores)
print("最も低いスコア:", q_min, "（", scores.index(q_min) + 1, "番目の塩基）")
print("不正確な確率:", round(10**(-q_min/10), 4))
````

### 実習1の答え合わせ

　生成AIが作ったプログラムの結果は、**本当に正しいか確認する**ことが大切です。まず、リード数を別の方法で確かめてみましょう。FASTQ形式では1リードが4行で書かれているので、ファイルの行数を4で割るとリード数になります。

````
%%bash
## wc -l でファイルの行数を数え、4で割ってリード数を求める
for fq in short_reads.R1.fastq short_reads.R2.fastq long_reads.fastq; do
    echo "$fq: $(( $(wc -l < $fq) / 4 )) reads"
done
````

　次のコードセルは、実習1で計算した値を、なるべく単純なプログラムで計算したものです。Geminiの結果と比べてみてください。

````
from Bio import SeqIO
from Bio.SeqUtils import gc_fraction
import statistics

for fq in ["short_reads.R1.fastq", "short_reads.R2.fastq", "long_reads.fastq"]:
    records = list(SeqIO.parse(fq, "fastq"))
    lengths = [len(r) for r in records]
    mean_q  = [statistics.mean(r.letter_annotations["phred_quality"]) for r in records]
    gc      = [gc_fraction(r.seq) * 100 for r in records]
    print(fq)
    print("  リード数        :", len(records))
    print("  リード長 平均   :", round(statistics.mean(lengths), 1))
    print("  リード長 中央値 :", statistics.median(lengths))
    print("  リード長 最小/最大:", min(lengths), "/", max(lengths))
    print("  平均品質スコア  :", round(statistics.mean(mean_q), 2))
    print("  平均GC含量 (%)  :", round(statistics.mean(gc), 2))
````

　Geminiの結果と一致しましたか？ 小数点以下などが少し違う場合は、**計算方法（定義）の違い**が原因かもしれません。たとえば「平均品質スコア」は、スコアをそのまま平均する方法と、不正確な確率に変換してから平均する方法があり、結果が変わります。生成AIを使うときは、どういう計算をしたのかをコードやAIの説明で確認する習慣をつけましょう。

#### もう一歩: クオリティスコアは本当に正確さを表している？

　今回のロングリードは、元のゲノム配列から擬似的に作られたデータなので、各リードが実際にどれくらい正確か（`read_identity`）がリード名の行に書かれています。クオリティスコアから計算した「予想される正確さ」と比べてみましょう。

````
from Bio import SeqIO
import statistics

print("リード長  平均Q  クオリティから予想した正確さ  実際の正確さ")
for r in list(SeqIO.parse("long_reads.fastq", "fastq"))[:10]:
    q = r.letter_annotations["phred_quality"]
    ## 各塩基の不正確な確率 10^(-Q/10) を平均し、1から引いて「正確さ」にする
    expected = 1 - statistics.mean([10**(-x/10) for x in q])
    actual   = r.description.split("read_identity=")[1]
    print(f"{len(r):>8}  {statistics.mean(q):5.1f}  {expected*100:>26.1f}%  {actual:>12}")
````

````
from Bio import Entrez, SeqIO
from Bio.SeqUtils import gc_fraction

Entrez.email = "your.name@example.com"  # ← 自分のメールアドレスに書き換える

accessions = {"ヒト": "NC_012920", "チンパンジー": "NC_001643", "イネ": "NC_011033"}

for name, acc in accessions.items():
    ## NCBIからFASTA形式で配列をダウンロードし、ファイルに保存する
    handle = Entrez.efetch(db="nucleotide", id=acc, rettype="fasta", retmode="text")
    record = SeqIO.read(handle, "fasta")
    handle.close()
    SeqIO.write(record, f"{acc}.fasta", "fasta")
    print(f"{name}（{acc}）: {len(record.seq):,} bp, GC含量 {gc_fraction(record.seq)*100:.1f}%")
````
